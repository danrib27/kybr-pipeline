## Exploring Licitações Dataset

```python
import polars as pl
```

```python
STR, FLOAT, INT = pl.String, pl.Float64, pl.Int64
```

```python
SCHEMA_EMPENHOS_RELACIONADOS = {
    "Número Licitação": STR,
    "Código UG": STR,
    "Nome UG": STR,
    "Código Modalidade Compra": STR,
    "Modalidade Compra": STR,
    "Número Processo": STR,
    "Código Empenho": STR,
    "Data Emissão Empenho": STR,
    "Observação Empenho": STR,
    "Valor Empenho (R$)": FLOAT,
}

df_er = pl.read_csv(
    "../data/extracted/licitacoes/202404_Licitacoes/202404_EmpenhosRelacionados.csv",
    encoding="latin-1",
    separator=";",
    schema=SCHEMA_EMPENHOS_RELACIONADOS,
    decimal_comma=True,
)
```

```python
df_er.describe()
```
