```python
import re
import unicodedata

import polars as pl
```

```python
STR, INT = pl.String, pl.Int64
DECIMAL_2, DECIMAL_4 = pl.Decimal(18, 2), pl.Decimal(18, 4)
```

```python
def to_snake_case(name: str) -> str:
    """Turn a source column name into a snake_case one.

    Turns "(R$)" into the "_brl" suffix, strips accents, lowercases
    and joins the words with "_", which also replaces any run of
    other characters, as " - ".
    """
    name = re.sub(r"\s*\(R\$\)", "_brl", name)
    name = unicodedata.normalize("NFKD", name)
    name = name.encode("ascii", "ignore").decode().lower()
    return re.sub(r"[^a-z0-9]+", "_", name).strip("_")
```
